# Training and testing on ATIS 

```python
!pip install -q datasets
from datasets import load_dataset

dataset = load_dataset("tuetschek/atis")

def to_rasa_format(text, slots):
    words = text.split()
    tags = slots.split()
    
    if len(words) != len(tags):
        return text 
    
    out = []
    i = 0
    while i < len(words):
        if tags[i].startswith('B-'):
            entity_name = tags[i][2:]
            span = [words[i]]
            i += 1
            while i < len(words) and tags[i] == f"I-{entity_name}":
                span.append(words[i])
                i += 1
            out.append(f"[{' '.join(span)}]({entity_name})")
        else:
            out.append(words[i])
            i += 1
    return " ".join(out)

def write_rasa_yaml(dataset_split, filepath):
    intent_dict = {}
    for row in dataset_split:
        
        intent = row['intent'].split('#')[0]
        rasa_text = to_rasa_format(row['text'], row['slots'])
        
        if intent not in intent_dict:
            intent_dict[intent] = []
        intent_dict[intent].append(rasa_text)
        
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write('version: "3.1"\n\n')
        f.write('nlu:\n')
        for intent, examples in intent_dict.items():
            f.write(f'- intent: {intent}\n')
            f.write('  examples: |\n')
            for ex in examples:
                f.write(f'    - {ex}\n')

os.makedirs("/root/myhomedir/bucket/DIET_torch/data", exist_ok=True)
write_rasa_yaml(dataset['train'], '/root/myhomedir/bucket/DIET_torch/data/atis_train.yml')
write_rasa_yaml(dataset['test'], '/root/myhomedir/bucket/DIET_torch/data/atis_test.yml')

print(f"Train : {len(dataset['train'])} sentences converted")
print(f"Test  : {len(dataset['test'])} sentences converted")
```
